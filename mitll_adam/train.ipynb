## Attack Training and Dataset Generation Code
The below code evaluates BERT, RoBERTa, and DistilBERT models on the IMDb dataset, and generates adversarial examples using the TextFooler attack recipe from the TextAttack library. The generated adversarial examples are saved in CSV files for further analysis (see `analyze_pairs.ipynb`).

### Config and Imports

```python
import os

import torch
from transformers import BertTokenizer, BertForSequenceClassification, Trainer, TrainingArguments, AutoTokenizer, AutoModelForSequenceClassification
from datasets import load_dataset
from textattack.models.wrappers import HuggingFaceModelWrapper
from textattack.attack_recipes import TextFoolerJin2019
from textattack.datasets import Dataset
from textattack import Attacker, AttackArgs
import nltk
```

```python
dataset = load_dataset('imdb')
# Tokenization function
def tokenize(batch, tokenizer):
    return tokenizer(batch['text'], padding='max_length', truncation=True, max_length=256)
```

## BERT

```python
!textattack train --model-name-or-path bert-base-uncased --dataset imdb --model-num-labels 2 --model-max-length 128 --per-device-train-batch-size 256 --num-epochs 8 --output-dir ./outputs/bert
```

```python
!textattack eval --num-examples 1000 --model ./outputs/bert/best_model/ --dataset-from-huggingface imdb --dataset-split test
```

```python
# Calculate mean and stdev num_words in imdb
from datasets import load_dataset
import pandas as pd

# Load dataset
dataset = load_dataset("imdb")

# Convert to pandas DataFrame
df = pd.concat([
    dataset["train"].to_pandas(),
    dataset["test"].to_pandas()
], ignore_index=True)

print(df.head())

# Count words per review
word_counts = df["text"].apply(lambda x: len(x.split()))

# Mean and standard deviation
mean_words = word_counts.mean()
std_words = word_counts.std()

print("Mean word count:", mean_words)
print("Standard deviation:", std_words)
```
